```python


np.random.seed(42)

departments = ["IT", "HR", "Finance", "Sales", "Marketing"]

df = pd.DataFrame({
    "Employee_ID": range(1001, 1051),
    "Department": np.random.choice(departments, 50),
    "Experience": np.random.randint(1, 16, 50),
    "Salary": np.random.randint(35000, 150000, 50),
    "Performance": np.random.randint(60, 100, 50),
    "Age": np.random.randint(22, 50, 50)
})

df.head()
```

```python
dep_salary = df.groupby('Department')['Salary'].mean().sort_values()
```

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.bar(dep_salary.index,
      dep_salary.values)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.barh(dep_salary.index,
      dep_salary.values)

plt.show()
```

```python
#box plot


np.random.seed(42)

departments = ["IT","HR","Finance","Sales","Marketing"]

df = pd.DataFrame({

    "Department": np.random.choice(departments,100),

    "Salary": np.random.normal(
        loc=70000,
        scale=15000,
        size=100
    ).astype(int),

    "Experience": np.random.randint(1,15,100),

    "Performance": np.random.randint(60,100,100)
})

df.head()
```

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.boxplot(df["Salary"])

ax.set_title("Salary Distribution")

ax.set_ylabel("Salary")

plt.show()
```

```python
salary_groups = [
    group["Salary"].values
    for _, group in df.groupby("Department")
]

labels = df["Department"].unique()

fig, ax = plt.subplots(figsize=(10,6))

ax.boxplot(salary_groups, labels=labels)

ax.set_title("Salary Distribution by Department")

plt.show()
```

```python
fig, ax = plt.subplots(2,2, figsize=(12,8))

ax[0,0].hist(df["Salary"])

ax[0,1].boxplot(df["Salary"])

ax[1,0].scatter(
    df["Experience"],
    df["Salary"]
)

dept = df["Department"].value_counts()

ax[1,1].bar(
    dept.index,
    dept.values
)

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(
    2,
    2,
    figsize=(14,10)
)

# Histogram
ax[0,0].hist(
    df["Salary"],
    bins=10,
    color="skyblue",
    edgecolor="black"
)

ax[0,0].set_title("Salary Distribution")

# Boxplot
ax[0,1].boxplot(df["Salary"])

ax[0,1].set_title("Salary Boxplot")

# Scatter
ax[1,0].scatter(
    df["Experience"],
    df["Salary"],
    alpha=0.7
)

ax[1,0].set_title("Experience vs Salary")

# Bar

dept = df["Department"].value_counts()

ax[1,1].bar(
    dept.index,
    dept.values,
    color="orange"
)

ax[1,1].set_title("Employees by Department")

plt.tight_layout()

plt.show()
```
